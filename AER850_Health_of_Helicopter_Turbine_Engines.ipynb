Looking for Correlations

```python
#import libraries
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import scatter_matrix
#importing data
X_train=pd.read_csv("/content/X_train.csv")
y_train=pd.read_csv("/content/y_train.csv")
train_data=X_train.merge(y_train, on="id")
corr_data=train_data.drop(columns=["id"])
corr_matrix=corr_data.corr(numeric_only=True)

print("Correlation with Faulty:")
print(corr_matrix["faulty"].sort_values(ascending=False))
print("\nCorrelation with Torque Margin:")
print(corr_matrix["trq_margin"].sort_values(ascending=False))
```

Correlation Heat Map

```python
import seaborn as sns

plt.figure(figsize=(10,10))
plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)

sns.heatmap(corr_matrix, annot=True, fmt=".2f",cmap="coolwarm",vmin=-1,vmax=1)
plt.title("Correlation Matrix of Engine Parameter")
plt.tight_layout()
plt.show()
```

Scatter matrix

```python
attributes=["faulty","trq_margin","pa","oat","mgt"]
scatter_matrix(train_data[attributes],figsize=(12,8), alpha=0.2)
plt.tight_layout()
plt.show()
```
